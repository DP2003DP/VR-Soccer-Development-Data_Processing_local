# Posting the EDF files to the box drive

This script allows posting to the box drive that contains all of the EEG data in EDF format

```python
import mne
import os
import numpy as np
import shutil
```

```python
# Put path to your output data folder here
username = os.getlogin()
gender = "Women"
type = "OnField"  # "OnField" or "VR"
edf_folder = f'C:\\Users\\{username}\\Box\\Research_SPORTS_VR_AI\\Creative Inquiry F24\\EDF Data\\{gender}\\{type}'
```

```python
# Find local directory with EDF files
epoch_edf_folder = f"{gender}_{type}_epochs_as_edf_by_annotation/"


# Create new directory to the BOX drive in the edf_folder directory
name = "Neely"
new_folder = f"{name}_Split_By_Annotation"

new_folder = os.path.join(edf_folder, new_folder)
os.makedirs(new_folder, exist_ok=True)
```

```python
# Find all EDF files from the specific name in the folder 

edf_files = [f for f in os.listdir(epoch_edf_folder) if f.endswith('.edf') and name in f]

# Add edf files to the new folder
for f in edf_files:
    src = os.path.join(epoch_edf_folder, f)
    dst = os.path.join(new_folder, f)
    shutil.copy(src, dst)
```
